```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR

# 🧩 Split data for first task: predict fuelConsumpCmb
X1 = cdf[['engineSize', 'cylinderNo']]
y1 = cdf['fuelConsumpCmb']

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
```

Linear Regression

```python
lr = LinearRegression()
lr.fit(X1_train, y1_train)
y1_pred_lr = lr.predict(X1_test)

mae_lr = mean_absolute_error(y1_test, y1_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y1_test, y1_pred_lr))
r2_lr = r2_score(y1_test, y1_pred_lr)

print("🔹 Linear Regression (fuelConsumpCmb)")
print(f"MAE: {mae_lr:.3f}")
print(f"RMSE: {rmse_lr:.3f}")
print(f"R²: {r2_lr:.3f}")
```

Decision Tree

```python
tree = DecisionTreeRegressor(random_state=42)
tree.fit(X1_train, y1_train)
y1_pred_tree = tree.predict(X1_test)

mae_tree = mean_absolute_error(y1_test, y1_pred_tree)
rmse_tree = np.sqrt(mean_squared_error(y1_test, y1_pred_tree))
r2_tree = r2_score(y1_test, y1_pred_tree)

print("🌳 Decision Tree (fuelConsumpCmb)")
print(f"MAE: {mae_tree:.3f}")
print(f"RMSE: {rmse_tree:.3f}")
print(f"R²: {r2_tree:.3f}")
```

Random Forest

```python
rf = RandomForestRegressor(random_state=42)
rf.fit(X1_train, y1_train)
y1_pred_rf = rf.predict(X1_test)

mae_rf = mean_absolute_error(y1_test, y1_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y1_test, y1_pred_rf))
r2_rf = r2_score(y1_test, y1_pred_rf)

print("🌲 Random Forest (fuelConsumpCmb)")
print(f"MAE: {mae_rf:.3f}")
print(f"RMSE: {rmse_rf:.3f}")
print(f"R²: {r2_rf:.3f}")
```

Gradient Boosting

```python
gb = GradientBoostingRegressor(random_state=42)
gb.fit(X1_train, y1_train)
y1_pred_gb = gb.predict(X1_test)

mae_gb = mean_absolute_error(y1_test, y1_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y1_test, y1_pred_gb))
r2_gb = r2_score(y1_test, y1_pred_gb)

print("⚡ Gradient Boosting (fuelConsumpCmb)")
print(f"MAE: {mae_gb:.3f}")
print(f"RMSE: {rmse_gb:.3f}")
print(f"R²: {r2_gb:.3f}")
```

Support Vector Regression (SVR)

```python
svr = SVR(kernel='rbf')
svr.fit(X1_train, y1_train)
y1_pred_svr = svr.predict(X1_test)

mae_svr = mean_absolute_error(y1_test, y1_pred_svr)
rmse_svr = np.sqrt(mean_squared_error(y1_test, y1_pred_svr))
r2_svr = r2_score(y1_test, y1_pred_svr)

print("💻 SVR (fuelConsumpCmb)")
print(f"MAE: {mae_svr:.3f}")
print(f"RMSE: {rmse_svr:.3f}")
print(f"R²: {r2_svr:.3f}")
```

```python
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest', 'Gradient Boosting', 'SVR'],
    'MAE': [mae_lr, mae_tree, mae_rf, mae_gb, mae_svr],
    'RMSE': [rmse_lr, rmse_tree, rmse_rf, rmse_gb, rmse_svr],
    'R²': [r2_lr, r2_tree, r2_rf, r2_gb, r2_svr]
})
print(results.sort_values(by='R²', ascending=False))
```

Exporting the models with joblib

```python
import os
import joblib
import numpy as np
from sklearn.ensemble import GradientBoostingRegressor

# Make sure fuelType is already numeric if it exists (not needed here, just a reminder)
assert np.issubdtype(cdf['engineSize'].dtype, np.number)
assert np.issubdtype(cdf['cylinderNo'].dtype, np.number)
assert np.issubdtype(cdf['fuelConsumpCmb'].dtype, np.number)

# Features & target
X_fc = cdf[['engineSize', 'cylinderNo']]
y_fc = cdf['fuelConsumpCmb']

# Train on FULL dataset (after you’ve selected the model via CV/test)
gb_fc = GradientBoostingRegressor(random_state=42)
gb_fc.fit(X_fc, y_fc)

# Export
os.makedirs("models", exist_ok=True)
joblib.dump(gb_fc, "models/gb_fuelconsump.pkl")

print("✅ Exported GradientBoosting model to models/gb_fuelconsump.pkl")
```

```python
import os
import joblib
import numpy as np
from sklearn.tree import DecisionTreeRegressor

# Ensure inputs are numeric
assert np.issubdtype(cdf['fuelConsumpCmb'].dtype, np.number)
assert np.issubdtype(cdf['fuelType'].dtype, np.number), "fuelType must be numeric (e.g., X→0, others→1)."

# Features & target
X_co2 = cdf[['fuelConsumpCmb', 'fuelType']]
y_co2 = cdf['co2Emissions']

# Train on FULL dataset
dt_co2 = DecisionTreeRegressor(random_state=42)
dt_co2.fit(X_co2, y_co2)

# Export
os.makedirs("models", exist_ok=True)
joblib.dump(dt_co2, "models/dt_co2.pkl")

print("✅ Exported DecisionTree model to models/dt_co2.pkl")
```
